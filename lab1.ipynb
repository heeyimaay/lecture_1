1. Проверка упорядоченности цифр справа налево


```python
def is_ordered_reverse(num):
    digits = [int(d) for d in str(num)]
    for i in range(len(digits) - 1):
        if digits[i] <= digits[i + 1]:
            return False
    return True

# Тестирование
number = int(input("Введите натуральное число:"))
if is_ordered_reverse(number):
    print("Цифры упорядочены по возрастанию справа налево")
else:
    print("Цифры НЕ упорядочены по возрастанию справа налево")
```

2. Анализ текста


```python
def analyze_text(text):
    vowels = "аеёиоуыэюяaeiou"
    consonants = "бвгджзйклмнпрстфхцчшщbcdfghjklmnpqrstvwxyz"

    vowel_count = 0
    consonant_count = 0
    consonant_letters = []

    for char in text.lower():
        if char in vowels:
            vowel_count += 1
        elif char in consonants:
            consonant_count += 1
            if char not in consonant_letters:
                consonant_letters.append(char)

    words_count = len(text.split())

    print(f"Гласных: {vowel_count}")
    print(f"Согласных: {consonant_count}")
    print(f"Количество слов: {words_count}")

    if vowel_count == consonant_count:
        print(f"Согласные буквы в тексте: {', '.join(consonant_letters)}")

# Тестирование
text = input("Введите текст: ")
analyze_text(text)
```

3. Подсчёт пар одинаковых элементов

```python
def count_equal_pairs(numbers):
    count = 0
    n = len(numbers)
    for i in range(n):
        for j in range(i + 1, n):
            if numbers[i] == numbers[j]:
                count += 1
    return count

# Тестирование
numbers = list(map(int, input("Введите числа через пробел: ").split()))
pairs_count = count_equal_pairs(numbers)
print(f"Количество пар одинаковых элементов: {pairs_count}")
```

4. Словарь частот цифр

```python
def create_frequency_dict(digit_string):
    freq_dict = {}
    for char in digit_string:
        if char.isdigit():
            num = int(char)
            freq_dict[num] = freq_dict.get(num, 0) + 1
    return freq_dict

# Тестирование
digit_sequence = input("Введите последовательность цифр: ")
result_dict = create_frequency_dict(digit_sequence)
print("Словарь частот:")
for key in sorted(result_dict.keys()):
    print(f"{key}: {result_dict[key]}")
```

5. Программа «Магазин»

```python
def shop_program():
    products = {
        "молоко": [80, 50],
        "хлеб": [40, 100],
        "яйца": [120, 30],
        "сыр": [300, 20],
        "колбаса": [250, 15]
    }

    while True:
        print("\n--- Меню Магазина ---")
        print("1. Просмотр цены")
        print("2. Просмотр количества")
        print("3. Вся информация")
        print("4. Покупка")
        print("5. До свидания")

        choice = input("Выберите пункт: ")

        if choice == '1':
            print("\n--- Цены товаров ---")
            for product, info in products.items():
                print(f"{product}: {info[0]} руб.")

        elif choice == '2':
            print("\n--- Количество товаров ---")
            for product, info in products.items():
                print(f"{product}: {info[1]} шт.")

        elif choice == '3':
            print("\n--- Вся информация ---")
            for product, info in products.items():
                print(f"{product}: цена {info[0]} руб., количество {info[1]} шт.")

        elif choice == '4':
            total_cost = 0
            purchased_items = {}

            while True:
                product_name = input("Введите название товара (или 'n' для завершения): ").strip().lower()
                if product_name == 'n':
                    break

                if product_name in products:
                    try:
                        quantity = int(input(f"Сколько '{product_name}' вы хотите купить? "))
                        if quantity <= 0:
                            print("Количество должно быть положительным!")
                            continue

                        available = products[product_name][1]
                        if quantity > available:
                            print(f"В наличии только {available} шт.")
                        else:
                            cost = products[product_name][0] * quantity
                            total_cost += cost
                            products[product_name][1] -= quantity
                            purchased_items[product_name] = purchased_items.get(product_name, 0) + quantity
                            print(f"Добавлено: {product_name} - {quantity} шт. ({cost} руб.)")
                    except ValueError:
                        print("Введите корректное число!")
                else:
                    print("Товар не найден!")

            print(f"\n--- Итог покупки ---")
            if purchased_items:
                for item, qty in purchased_items.items():
                    print(f"{item}: {qty} шт.")
                print(f"Общая стоимость: {total_cost} руб.")
            else:
                print("Вы ничего не купили")

            print("\n--- Остатки товаров ---")
            for product, info in products.items():
                print(f"{product}: {info[1]} шт.")

        elif choice == '5':
            print("До свидания!")
            break

        else:
            print("Неверный выбор!")

# Запуск программы
shop_program()
```

6. Кортеж уникальных элементов в обратном порядке

```python
def unique_reverse_tuple(numbers):
    unique_nums = []
    for num in numbers:
        if num not in unique_nums:
            unique_nums.append(num)
    return tuple(unique_nums[::-1])

# Тестирование
numbers = list(map(int, input("Введите целые числа через пробел: ").split()))
result = unique_reverse_tuple(numbers)
print(f"Кортеж уникальных элементов в обратном порядке: {result}")
```

1. Функция проверки простого числа

```python
def is_prime(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

# Тестирование
num = int(input("Введите число от 0 до 1000: "))
if 0 <= num <= 1000:
    if is_prime(num):
        print(f"{num} - простое число")
    else:
        print(f"{num} - составное число")
else:
    print("Число должно быть от 0 до 1000")
```

2. Универсальная функция обработки разных типов данных

```python
def process_data(data):
    if isinstance(data, list):
        # Список: посчитать длину всех его слов
        total_length = 0
        words_count = 0
        for item in data:
            if isinstance(item, str):
                total_length += len(item)
                words_count += 1
        return f"Список: найдено {words_count} слов, общая длина = {total_length}"

    elif isinstance(data, tuple):
        # Кортеж: посчитать кол-во букв и чисел
        letters_count = 0
        numbers_count = 0
        for item in data:
            if isinstance(item, str):
                letters_count += len([char for char in item if char.isalpha()])
            elif isinstance(item, (int, float)):
                numbers_count += 1
        return f"Кортеж: букв = {letters_count}, чисел = {numbers_count}"

    elif isinstance(data, int):
        # Число: вывести в обратном порядке
        reversed_num = int(str(data)[::-1])
        return f"Число {data} в обратном порядке: {reversed_num}"

    elif isinstance(data, str):
        # Строка: посчитать сумму всех цифр
        digits_sum = sum(int(char) for char in data if char.isdigit())
        return f"Строка: сумма всех цифр = {digits_sum}"

    else:
        return "Неподдерживаемый тип данных"

# Тестирование всех случаев
print("--- Тестирование функции process_data ---")

# Список
test_list = ["hello", "world", 123, "python", "programming"]
print(f"Вход: {test_list}")
print(f"Результат: {process_data(test_list)}\n")

# Кортеж
test_tuple = ("apple", 123, "banana", 45.6, "cat")
print(f"Вход: {test_tuple}")
print(f"Результат: {process_data(test_tuple)}\n")

# Число
test_number = 12345
print(f"Вход: {test_number}")
print(f"Результат: {process_data(test_number)}\n")

# Строка
test_string = "В 2024 году было 365 дней и 12 месяцев"
print(f"Вход: '{test_string}'")
print(f"Результат: {process_data(test_string)}\n")
```

3. Проверка симметричности матрицы

```python
def is_symmetric_matrix(matrix):
    n = len(matrix)

    # Проверяем, является ли матрица квадратной
    for row in matrix:
        if len(row) != n:
            return False

    # Проверяем симметричность
    for i in range(n):
        for j in range(i + 1, n):
            if matrix[i][j] != matrix[j][i]:
                return False
    return True

# Ввод матрицы
n = int(input("Введите размер квадратной матрицы: "))
print("Введите матрицу построчно (числа через пробел):")
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)

# Проверка и вывод
print("\nВведенная матрица:")
for row in matrix:
    print(row)

if is_symmetric_matrix(matrix):
    print("Матрица является симметричной")
else:
    print("Матрица НЕ является симметричной")
```

Задание 1: Класс String

```python
class String:
    def __init__(self, text=""):
        self.text = str(text)

    def __str__(self):
        return self.text

    def length(self):
        """Возвращает длину строки"""
        return len(self.text)

    def reverse(self):
        """Возвращает перевернутую строку"""
        return String(self.text[::-1])

    def count_words(self):
        """Подсчитывает количество слов в строке"""
        words = self.text.split()
        return len(words)

    def is_palindrome(self):
        """Проверяет, является ли строка палиндромом"""
        clean_text = ''.join(char.lower() for char in self.text if char.isalnum())
        return clean_text == clean_text[::-1]

    def find_substring(self, substring):
        """Находит первое вхождение подстроки"""
        return self.text.find(substring)

    def replace(self, old, new):
        """Заменяет все вхождения подстроки"""
        return String(self.text.replace(old, new))

    def to_upper(self):
        """Преобразует строку в верхний регистр"""
        return String(self.text.upper())

    def to_lower(self):
        """Преобразует строку в нижний регистр"""
        return String(self.text.lower())

# Демонстрация работы класса
print("--- Демонстрация класса String ---")
s = String("Hello World! А роза упала на лапу Азора")

print(f"Исходная строка: {s}")
print(f"Длина: {s.length()}")
print(f"Перевернутая: {s.reverse()}")
print(f"Количество слов: {s.count_words()}")
print(f"Является палиндромом: {s.is_palindrome()}")
print(f"Поиск 'World': {s.find_substring('World')}")
print(f"Замена 'Hello' на 'Hi': {s.replace('Hello', 'Hi')}")
print(f"Верхний регистр: {s.to_upper()}")
print(f"Нижний регистр: {s.to_lower()}")
```

Задание 2: Класс Country

```python
class Country:
    def __init__(self, name, capital, area, population):
        self.name = name
        self.capital = capital
        self.area = area  # площадь в тыс. км²
        self.population = population  # население в млн

    def __str__(self):
        return f"{self.name} (столица: {self.capital}, площадь: {self.area} тыс.км², население: {self.population} млн)"

    def __repr__(self):
        return f"Country('{self.name}', '{self.capital}', {self.area}, {self.population})"

def filter_countries_by_area(countries, min_area):
    """Фильтрует страны по минимальной площади"""
    return [country for country in countries if country.area >= min_area]

def filter_countries_by_population(countries, min_population):
    """Фильтрует страны по минимальному населению"""
    return [country for country in countries if country.population >= min_population]

# Создание списка стран
countries = [
    Country("Россия", "Москва", 17125, 146),
    Country("Канада", "Оттава", 9985, 38),
    Country("Китай", "Пекин", 9597, 1402),
    Country("США", "Вашингтон", 9834, 331),
    Country("Бразилия", "Бразилиа", 8515, 214),
    Country("Австралия", "Канберра", 7692, 26),
    Country("Индия", "Нью-Дели", 3287, 1380),
    Country("Аргентина", "Буэнос-Айрес", 2780, 45)
]

# Демонстрация работы
print("\n--- Демонстрация класса Country ---")
print("Все страны:")
for country in countries:
    print(f"  - {country}")

# Фильтрация по площади
print(f"\na) Страны с площадью более 5000 тыс.км²:")
large_countries = filter_countries_by_area(countries, 5000)
for country in large_countries:
    print(f"  - {country}")

# Фильтрация по населению
print(f"\nb) Страны с населением более 100 млн:")
populous_countries = filter_countries_by_population(countries, 100)
for country in populous_countries:
    print(f"  - {country}")
```

Задание 3: Классы Зоомагазина

```python
class Animal:
    def __init__(self, breed, price):
        self.breed = breed
        self.price = price

    def move(self):
        return "Передвигается"

    def __str__(self):
        return f"{self.breed} - {self.price} руб."

class Fish(Animal):
    def __init__(self, breed, price, water_type="пресная"):
        super().__init__(breed, price)
        self.water_type = water_type

    def move(self):
        return "Плавает в воде"

    def __str__(self):
        return f"Рыба: {self.breed} ({self.water_type} вода) - {self.price} руб."

class Bird(Animal):
    def __init__(self, breed, price, can_fly=True):
        super().__init__(breed, price)
        self.can_fly = can_fly

    def move(self):
        if self.can_fly:
            return "Летает в небе"
        else:
            return "Ходит по земле"

    def __str__(self):
        flight = "летающая" if self.can_fly else "нелетающая"
        return f"Птица: {self.breed} ({flight}) - {self.price} руб."

class PetShop:
    def __init__(self, name):
        self.name = name
        self.animals = []

    def add_animal(self, animal):
        self.animals.append(animal)

    def get_most_expensive(self):
        if not self.animals:
            return None
        return max(self.animals, key=lambda x: x.price)

    def save_to_file(self, filename):
        with open(filename, 'w', encoding='utf-8') as f:
            f.write(f"Каталог зоомагазина '{self.name}'\n")
            f.write("=" * 40 + "\n")

            most_expensive = self.get_most_expensive()
            if most_expensive:
                f.write(f"Самая дорогая порода: {most_expensive}\n")

            f.write("\nВсе животные:\n")
            for animal in self.animals:
                f.write(f"- {animal} -> {animal.move()}\n")

        print(f"Информация сохранена в файл: {filename}")

# Демонстрация работы
print("\n--- Демонстрация зоомагазина ---")
shop = PetShop("Дружок")

# Добавляем животных
shop.add_animal(Fish("Золотая рыбка", 500))
shop.add_animal(Fish("Скалярия", 1200, "пресная"))
shop.add_animal(Fish("Клоун", 2500, "морская"))
shop.add_animal(Bird("Попугай ара", 15000))
shop.add_animal(Bird("Канарейка", 3000))
shop.add_animal(Bird("Страус", 50000, False))
shop.add_animal(Bird("Пингвин", 45000, False))

# Выводим информацию
print("Все животные в магазине:")
for animal in shop.animals:
    print(f"  - {animal} -> {animal.move()}")

# Самая дорогая порода
most_expensive = shop.get_most_expensive()
print(f"\nСамая дорогая порода: {most_expensive}")

# Сохраняем в файл
shop.save_to_file("pet_shop_catalog.txt")
```

Задание 4: Собственный класс (Банковский счет)

```python
class BankAccount:
    BANK_NAME = "Национальный Банк"
    accounts_count = 0

    def __init__(self, owner_name, initial_balance=0):
        self.owner = owner_name
        self._balance = initial_balance
        self.account_number = BankAccount.accounts_count + 1000
        BankAccount.accounts_count += 1

    # Метод экземпляра
    def deposit(self, amount):
        """Пополнение счета"""
        if amount > 0:
            self._balance += amount
            print(f"Счет пополнен на {amount} руб. Новый баланс: {self._balance} руб.")
        else:
            print("Сумма пополнения должна быть положительной")

    # Метод экземпляра
    def withdraw(self, amount):
        """Снятие со счета"""
        if amount > 0:
            if amount <= self._balance:
                self._balance -= amount
                print(f"Со счета снято {amount} руб. Новый баланс: {self._balance} руб.")
                return True
            else:
                print("Недостаточно средств на счете")
                return False
        else:
            print("Сумма снятия должна быть положительной")
            return False

    # Метод экземпляра
    def get_balance(self):
        """Получение текущего баланса"""
        return self._balance

    # Метод класса
    @classmethod
    def get_bank_info(cls):
        """Информация о банке"""
        return f"Банк: {cls.BANK_NAME}, всего счетов: {cls.accounts_count}"

    # Статический метод
    @staticmethod
    def validate_amount(amount):
        """Проверка корректности суммы"""
        return isinstance(amount, (int, float)) and amount > 0

    @staticmethod
    def convert_currency(amount, rate):
        """Конвертация валюты"""
        return amount * rate

    def __str__(self):
        return f"Счет №{self.account_number}, владелец: {self.owner}, баланс: {self._balance} руб."

# Демонстрация работы
print("\n--- Демонстрация класса BankAccount ---")

# Создание счетов
account1 = BankAccount("Иван Иванов", 1000)
account2 = BankAccount("Петр Петров", 500)

print(account1)
print(account2)

# Использование методов экземпляра
print("\n--- Операции со счетом ---")
account1.deposit(500)
account1.withdraw(200)
account1.withdraw(2000)  # Недостаточно средств

# Использование методов класса
print(f"\n--- Информация о банке ---")
print(BankAccount.get_bank_info())

# Использование статических методов
print(f"\n--- Статические методы ---")
print(f"Проверка суммы 1000: {BankAccount.validate_amount(1000)}")
print(f"Проверка суммы -100: {BankAccount.validate_amount(-100)}")
print(f"Конвертация 100 USD в RUB (курс 90): {BankAccount.convert_currency(100, 90)} руб.")

# Создание еще одного счета для демонстрации счетчика
account3 = BankAccount("Сергей Сергеев")
print(f"\nПосле создания третьего счета:")
print(BankAccount.get_bank_info())
```
